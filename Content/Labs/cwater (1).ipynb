# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie


De soortelijke warmte $c$ van een stof geeft aan hoeveel energie nodig is om 1 kg van die stof 1 °C in temperatuur te laten stijgen. Deze grootheid wordt beschreven door:

$$
Q = m c \Delta T 
$$

In dit experiment wordt het water verwarmd met een elektrisch element. De elektrische energie is:
$$
Q = P t 
$$

Door $Q$ en $\Delta T$ te meten kan $c$ worden bepaald via:
$$
c = \frac{Q}{m \Delta T} 
$$

# Resultaten


````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

m_maatbeker = 0.2766  # in kg
m_totaal = 0.7437     # in kg
m_water = m_totaal - m_maatbeker  # in kg

t = np.array([0,20,40,60,80,100,120,140,160,180,200,220,240,260,280,300,320,340,360,380,400,420])  # in s
T = np.array([20.6,20.7,20.9,21.1,21.3,21.5,21.8,22.0,22.3,22.6,22.9,23.2,23.5,23.7,24.0,24.2,24.4,24.7,24.9,25.2,25.4,25.6])  # in °C

P = 25  # gebruikt vermogen in Watt

def lineair(t, a, b):
    return a * t + b

popt, pcov = curve_fit(lineair, t, T)
a, b = popt

T_fit = lineair(t, a, b)

t_tot = t[-1] - t[0]                 # totale tijd
Q = P * t_tot                        # toegevoerde energie in J
delta_T = T[-1] - T[0]               # temperatuurstijging in °C

c_water = Q / (m_water * delta_T)    # soortelijke warmte in J/(kg·K)

print(f"Soortelijke warmte van water: c = {c_water:.0f} J/(kg·K)")
````

````python
plt.figure()
plt.plot(t, T, 'o', label='Metingen')
plt.plot(t, T_fit, '-', label='Lineaire fit')
plt.xlabel("Tijd (s)")
plt.ylabel("Temperatuur (°C)")
plt.legend()
plt.grid(True)

# plt.savefig("figures/temperatuur_vs_tijd.png", dpi=450)
plt.show()
````
